# Linear Regression: Gradient Descent & Scikit-Learn

We want to find a straight line, **$y = m \cdot x + b$**, that best fits the data:
- $X = [1, 2, 3, 4, 5]$
- $Y = [2, 3, 4, 5, 6]$

Here, **$m$** is the slope and **$b$** is the intercept.

### Step 1: Import Libraries and Setup Data

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Example data
X = np.array([1, 2, 3, 4, 5], dtype=float)
Y = np.array([2, 3, 4, 5, 6], dtype=float)

print("X data:", X)
print("Y data:", Y)
```

---
### Step 2: Method 1 - Gradient Descent from Scratch

```python
# Hyperparameters
learning_rate = 0.01
epochs = 1000

# Start with slope (m) and intercept (b) as 0
m = 0.0
b = 0.0
n = len(X)

for i in range(epochs):
    # Predict
    Y_pred = m * X + b

    # Mean squared error (Cost)
    cost = ((Y_pred - Y) ** 2).mean()

    # Derivatives
    derivative_m = (-2 / n) * np.sum(X * (Y - Y_pred))
    derivative_b = (-2 / n) * np.sum(Y - Y_pred)

    # Update parameters
    m -= learning_rate * derivative_m
    b -= learning_rate * derivative_b

    if (i + 1) % 100 == 0:
        print(f"Epoch {i + 1}/{epochs}, Cost: {cost:.4f}, m: {m:.4f}, b: {b:.4f}")

print(f"\nFinal learned line: y = {m:.4f} * x + {b:.4f}")

# Prediction for new inputs
new_x = 6
prediction = m * new_x + b
print(f"Prediction for x={new_x}: {prediction:.4f}")

new_x1 = 7
prediction1 = m * new_x1 + b
print(f"Prediction for x={new_x1}: {prediction1:.4f}")
```

---
### Step 3: Method 2 - Scikit-Learn `LinearRegression`

```python
# Scikit-Learn expects 2D array for X
X_reshaped = X.reshape(-1, 1)

# Train model
model = LinearRegression()
model.fit(X_reshaped, Y)

m_sk = model.coef_[0]
b_sk = model.intercept_

print(f"Slope (m): {m_sk:.4f}")
print(f"Intercept (b): {b_sk:.4f}")
print(f"Learned line: y = {m_sk:.4f} * x + {b_sk:.4f}")

# Predictions for x=6 and x=7
print(f"Prediction for x=6: {model.predict([[6]])[0]:.4f}")
print(f"Prediction for x=7: {model.predict([[7]])[0]:.4f}")
```

---
### Step 4: Matplotlib Best Fit Line (Scikit-Learn Only)

```python
# Plot data points
plt.scatter(X, Y, color='blue', label='Actual Data')

# Plot best fit line for scikit-learn only
plt.plot(X, model.predict(X_reshaped), color='red', label=f'Best Fit Line: y = {m_sk:.2f}x + {b_sk:.2f}')

plt.xlabel('X')
plt.ylabel('Y')
plt.title('Linear Regression Best Fit Line (Scikit-Learn)')
plt.legend()
plt.grid(True)
plt.show()
```
